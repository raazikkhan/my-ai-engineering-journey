### Array Operations


```python
import numpy as np
arr1 = np.arange(12).reshape(3,4)
arr2 = np.arange(12,24).reshape(3,4)
print(arr1, arr2)
```

```python
# Scalar operations is used to perform operations on each element of the array.

#Arithmetic operations 

arr1 + 2
arr1 - 2
arr1 ** 3
```

```python
# Relational Operations

arr1 == arr1
```

```python
# Vector Operations we can perform operations on two arrays of the same shape. The operation is performed element-wise.
arr1 + arr2
arr1 - arr2
arr1 * arr2
arr1 / arr2

arr1 ** arr2
```

### Function in Numpy


```python
fun1 = np.random.random((3,3))
fun2 = np.round(fun1 * 100)
fun2
```

```python
# Min, Max, Product (0 = col, 1 = row)

np.max(arr1)

np.min(arr2)

np.prod(arr1, axis=0)
```

```python
# mean, median, standard deviation, variance

# Both are measures of central tendency used in statistics to find the center or typical value of a dataset
np.mean(arr2)
np.median(arr1)
# square root of the variance
std = np.arange(5)
np.std(std)
# how far apart a set of numbers are spread out from their average (mean) value
np.var(arr1)



```

```python
# Dot Product is used to multiply two arrays together. The dot product of two arrays is the sum of the products of their corresponding elements.
# important
a1 = np.arange(12).reshape(3,4)
a2 = np.arange(12,24).reshape(4,3)
a3 = np.dot(a1,a2)
a3
```

```python
#log and exponents

# log is used to calculate the natural logarithm of each element in the array. The natural logarithm is the logarithm to the base e, where e is the base of the natural logarithm (approximately equal to 2.71828).
np.log(a1)

# exponential function is used to calculate the exponential of each element in the array. The exponential function is defined as e^x, where e is the base of the natural logarithm (approximately equal to 2.71828).
np.exp(a1)
```

```python
# round, floor, ceil

# round is used to round each element in the array to the nearest integer.
np.round(a1)

# floor used to round down to the nearest integer less than or equal to each element in the array.
np.floor(a1)
 # ceil used to round up to the nearest integer greater than or equal to each element in the array.
np.ceil(a2)
```

### Reshaping


```python
# reshape, Transpose, ravel

# reshape
a = np.arange(12).reshape(3,4)
a
```

```python
# Transpose used to swap the rows and columns of a 2D array. It returns a new array with the same data but with the axes transposed.

a.T
```

```python
# Ravel used to flatten a multi-dimensional array into a one-dimensional array. It returns a new array that contains all the elements of the original array in a single dimension.

a.ravel()
```

### Stacking


```python
# horizontal stacking (hstack) used to combine two or more arrays horizontally (i.e., along the second axis). It takes a sequence of arrays as input and returns a new array that contains all the input arrays stacked side by side.

a_horizontal = np.hstack((arr1, arr2))
a_horizontal
```

```python
# Vertical stacking (vstack) used to combine two or more arrays vertically (i.e., along the first axis). It takes a sequence of arrays as input and returns a new array that contains all the input arrays stacked on top of each other.
a_vertical = np.vstack((arr1, arr2))
a_vertical
```

### Splitting  

```python
# Horizontal splitting (hsplit) used to split an array into multiple sub-arrays horizontally (i.e., along the second axis). It takes an array and the number of splits as input and returns a list of sub-arrays.
a_hsplit = np.hsplit(a_horizontal, 2)
a_hsplit
```

```python
# Virtical splitting (vsplit) used to split an array into multiple sub-arrays vertically (i.e., along the first axis). It takes an array and the number of splits as input and returns a list of sub-arrays.
a_vsplit = np.vsplit(a_vertical, 2)
a_vsplit
```
