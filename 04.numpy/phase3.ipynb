### Python List v/s Numpy Array

```python
# Python List
# Speed

pyList1 = [i for i in range(10000000)]
pyList2 = [i for i in range(10000000,20000000)]

pyList3 = []

import time
start = time.time()
for i in range(len(pyList1)):
    pyList3.append(pyList1[i] + pyList2[i])

end = time.time()
print(f"Time taken: {end - start} seconds")
```

```python
# Numpy Array
import numpy as np
import time
start = time.time()
npArr1 = np.arange(10000000)
npArr2 = np.arange(10000000,20000000)

npArr3 = npArr1 + npArr2
end = time.time()
print(f"Time Taken: {end - start} seconds")
```

```python
# memory usage in python
import sys
a = [i for i in range(10000)]
sys.getsizeof(a)
```

```python
# We Can reduce memory usage by using numpy arrays instead of python lists. Numpy arrays are more memory efficient than python lists because they store data in a contiguous block of memory, which allows for better cache performance and less overhead.
a = np.arange(10000, dtype=np.int32)
sys.getsizeof(a)
```

###  Advance Indexing


```python
# Normal Indexing and Slicing
idx = np.arange(12).reshape(4,3)
idx
```

```python
idx[1,2]  # Normal Indexing
```

```python
idx[2:, 1:] # Slicing
```

```python
# Fancy Indexing
arr = np.arange(12).reshape(4,3)
arr
```

```python
arr[[0,2,3], :] # fancy indexing allows us to select multiple specific elements using index arrays or Boolean conditions. It is useful for filtering, extracting, and manipulating data efficiently.
```

```python
# Boolean Indexing or Masking 

# Basic 
# remember - True = keep, false = ignore

arr = np.array([30,55,42,2,44,12,44,32,88])
arr > 30
arr[arr > 30] # Masking automatically  remove false value from the array 
```

```python
arr[arr > 25]
```

```python
arr[arr == 55]
```

```python
# Suppose you want numbers that are greater than 20 AND less than 50.
arr[(arr > 30) & (arr < 50)]
```

### In NumPy, use &, |, and ~ for array conditions. Wrap each condition in parentheses. Don't use Python's and or or with NumPy arrays.

```python
# Select numbers below 20 or above 50
arr[(arr < 20) | (arr > 50)]
```

```python
#Advance 
arr2D = np.random.randint(1,100,24).reshape(6,4)
arr2D
```

```python
# select all numbers greater then 30
arr2D[arr2D > 30]
```

```python
# Find all even numbers 
arr2D[arr2D % 2 == 0]
```

```python
# find all number greater then 50 & even
arr2D[(arr2D > 50) & (arr2D % 2 == 0)]
```

```python
# find all number divisible by 6
arr2D[arr2D % 7 == 0]
```

```python
# Select everything that is not greater than 30:
arr2D[~(arr2D > 30)]
```

### Broadcasting

```python
# Broadcasting with different arrays
a = np.arange(12).reshape(4,3)
b = np.arange(3)
print(a)
print(b)
print(a+b)
```

```python
# Why do we need broadcasting?
# Normally, we might want to add a single number to every element of an array.
# NumPy automatically applies 5 to every element. This is a simple form of broadcasting.
```
