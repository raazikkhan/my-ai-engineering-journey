# Sigmoid 

```python
# Sigmoid is a mathematical function that converts any real-valued number into a value between 0 and 1.
# Remember it simply: Raw number → Sigmoid → value between 0 and 1.

# Where is Sigmoid used?
# Logistic Regression – converts model scores into probabilities.
# Neural Networks – can be used as an activation function.
# Binary Classification – problems with two outcomes.
# Probability estimation – outputs values in the range 0–1.

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

values = np.arange(80)

print(sigmoid(values))
```

# Mean Squared Error 
### Mean Squared Error (MSE) is a loss function that measures how far a model's predictions are from the actual values.

```python
# Remember it: Actual − Predicted → Square → Average.
# When is MSE used?
# MSE is commonly used for regression problems, where the model predicts continuous numerical values.

# House price prediction
# Predicting temperature
# Sales forecasting
# Predicting salary
# Machine learning model training

actual = np.random.randint(1, 50, 25)

actual
```

```python
predicted = np.random.randint(1,50,25)

predicted
```

```python


def mse(actual,predicted):
    return np.mean((actual - predicted)** 2)
mse(actual, predicted)



# mse = np.mean((actual - predicted)** 2)
# print(f"Mean Squared Error is: {mse}")


# Here is what happens:
# actual - predicted → calculates the errors.
# . ** 2 → squares every error
# np.mean() → calculates the average squared error.
```
